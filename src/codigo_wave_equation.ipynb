# Imports

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import math
import scipy
from pandas import DataFrame, Series
```

# Datos de entrada

```python
c = math.sqrt(10e5)
a = 1 #Longitud de la cuerda
h = 1 #Altura
```

```python
time = [0, 0.3 * a / c, 0.5 * a / c, 0.7 * a / c, a / c]
time_legend = ["0", "0.3 * a/c", "0.5 * a/c","0.7 * a/c","a/c"]
```

# Funciones

```python
def lambda_n(n):
    lam = np.pi * n / a
    return lam
```

```python
def a_n(n):
    an = np.sin(np.pi * n / 2) * 8 * h / (a * lambda_n(n)) ** 2
    return an
```

```python
def u(x,t):
    sumatoria = 0
    for i in range(1,5):
        sumatoria += a_n(i) * (np.sin(lambda_n(i) * (x - c *t)) + np.sin(lambda_n(i) * (x + c * t)))
    sumatoria = sumatoria / 2
    return sumatoria
```

```python
def f1(x,t):
    sumatoria = 0
    for i in range(1,5):
        sumatoria += a_n(i) * (np.sin(lambda_n(i) * (x - c *t)))
    sumatoria = sumatoria / 2
    return sumatoria
```

```python
def f2(x,t):
    sumatoria = 0
    for i in range(1,5):
        sumatoria += a_n(i) * (np.sin(lambda_n(i) * (x + c *t)))
    sumatoria = sumatoria / 2
    return sumatoria
```

```python
x = np.linspace(0,a,100)
```

```python
for j in range(len(time)): 
    desplazamiento = np.zeros_like(x)
    funcion1 = np.zeros_like(x)
    funcion2 = np.zeros_like(x)
    for i in range(len(x)):
        desplazamiento[i] = u(x[i],time[j])
        funcion1[i] = f1(x[i],time[j])
        funcion2[i] = f2(x[i],time[j])
    plt.plot(x,desplazamiento, label =f" $t = {time_legend[j]}$")
    #plt.plot(x,funcion1, label =f" $t = {time_legend[j]}$")
    #plt.plot(x,funcion2, label =f" $t = {time_legend[j]}$")
    plt.xlabel("$x$")
    plt.ylabel(r"$u(x,t)$")
    #plt.legend()
    plt.tight_layout()
    plt.grid(True)
    plt.title("Tarea 3.9.1. Ecuación de onda")
    
```
